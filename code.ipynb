Mengidentifikasi pola jeda waktu antara perubahan harga pada masing masing indikator transmisi terhadap pergerakan harga beras domestik.

Expected output: Profil transmisi harga berdasarkan periode, estimasi respons dinamis jika didukung model, dan visualisasi jeda waktu yang mudah dipahami.

**Objective 2 - External Shock & Asymmetry Analysis**

Mengevaluasi apakah hubungan antara indikator eksternal dan harga beras domestik menunjukkan pola yang berbeda pada kondisi pasar yang berbeda, termasuk ketika risiko geopolitik meningkat atau rupiah mengalami depresiasi.

Analisis juga mengeksplorasi kemungkinan adanya respons yang tidak simetris terhadap kenaikan dan penurunan harga acuan internasional.

**Objective 3 - Early Warning Indicator Exploration**

Mengeksplorasi apakah kombinasi indikator harga internasional, nilai tukar, dan risiko geopolitik dapat memberikan informasi tambahan untuk memantau potensi tekanan harga beras domestik pada periode mendatang.

**Dataset Harga Beras Domestik**

*Source*: https://www.bps.go.id/id/statistics-table/2/Mjk1IzI=/rata-rata-harga-beras-di-tingkat-perdagangan-besar-grosir-indonesia.html

```python
file_path = '/content/drive/MyDrive/Colab Notebooks/'

def process_domestic_rice(nama_file):
  full_path = file_path + "Average Wholesale Rice Price at the National Level/"+ nama_file
  year = int(re.search(r'\d{4}',nama_file).group())

  month_map = {
        'Januari': 1, 'Februari': 2, 'Maret': 3, 'April': 4,
        'Mei': 5, 'Juni': 6, 'Juli': 7, 'Agustus': 8,
        'September': 9, 'Oktober': 10, 'November': 11, 'Desember': 12
  }

  month_eng = {
    'Januari': 'January',
    'Februari': 'February',
    'Maret': 'March',
    'April': 'April',
    'Mei': 'May',
    'Juni': 'June',
    'Juli': 'July',
    'Agustus': 'August',
    'September': 'September',
    'Oktober': 'October',
    'November': 'November',
    'Desember': 'December'
  }

  if nama_file.endswith('.csv'):
        df_domestic_rice_raw = pd.read_csv(full_path, skiprows=3)
  else:
        df_domestic_rice_raw = pd.read_excel(full_path, skiprows=3)

  df_domestic_rice = (
        df_domestic_rice_raw.rename(columns=lambda x:'Commodity' if 'Unnamed' in str (x) else x)
        .iloc[[0]].drop(columns=['Tahunan'], errors='ignore')
        .melt(id_vars=['Commodity'], var_name='Month', value_name='Price_Raw')
        .assign(
            Year=year,
            Month_number=lambda x: x['Month'].map(month_map),
            Month=lambda x: x['Month'].replace(month_eng),
            Domestic_Price=lambda x: pd.to_numeric(
                x['Price_Raw'], errors='coerce'
            )
        )
        .dropna(subset=['Domestic_Price'])
        .sort_values(by=['Year','Month_number'])
        .reset_index(drop=True)
        [['Year','Month','Domestic_Price']]
    )
  return df_domestic_rice

files = [
    "Data Harga Beras Grosir Nasional 2010.xlsx",
    "Data Harga Beras Grosir Nasional 2011.xlsx",
    "Data Harga Beras Grosir Nasional 2012.xlsx",
    "Data Harga Beras Grosir Nasional 2013.csv",
    "Data Harga Beras Grosir Nasional 2014.xlsx",
    "Data Harga Beras Grosir Nasional 2015.xlsx",
    "Data Harga Beras Grosir Nasional 2016.xlsx",
    "Data Harga Beras Grosir Nasional 2017.xlsx",
    "Data Harga Beras Grosir Nasional 2018.xlsx",
    "Data Harga Beras Grosir Nasional 2019.xlsx",
    "Data Harga Beras Grosir Nasional 2020.xlsx",
    "Data Harga Beras Grosir Nasional 2021.xlsx",
    "Data Harga Beras Grosir Nasional 2022.xlsx",
    "Data Harga Beras Grosir Nasional 2023.xlsx",
    "Data Harga Beras Grosir Nasional 2024.xlsx",
    "Data Harga Beras Grosir Nasional 2025.xlsx",
    "Data Harga Beras Grosir Nasional 2026.xlsx"
]

df_domestic_rice_list = pd.concat([process_domestic_rice(file) for file in files], ignore_index= True)

print("=== Dataset Information ==")
df_domestic_rice_list.info()

print("=== Dataset Head ==")
df_domestic_rice_list.head()
```

**Dataset Geopolitic Risk Index**

*Source* = https://www.matteoiacoviello.com/gpr.htm

```python
def process_gpr(file_path):
  column=['month','GPR']

  df_geopolitic_raw = (
      pd.read_excel(file_path, usecols=column)
      .assign(
          month=lambda x:pd.to_datetime(
              x['month'], format = '%B %Y'
          )

        )
      .loc[lambda x: x['month'].dt.year >= 2010]
      .assign(
            Year=lambda x: x['month'].dt.year,
            Month=lambda x: x['month'].dt.month_name()
      )
     .dropna(subset = ['month'])
      .sort_values(by='month')
      .reset_index(drop=True)
      [['Year', 'Month', 'GPR']]
      )

  return df_geopolitic_raw

df_geopolitic = process_gpr(file_path + "GPR.xls")

print("=== Dataset Information ==")
df_geopolitic.info()

print("=== Dataset Head ==")
df_geopolitic.head()
```

**Global Oil Price**

```python

def process_Global_Oil_Price(file_path, sheet_name=0):
    month_map = {
        1: 'January',
        2: 'February',
        3: 'March',
        4: 'April',
        5: 'May',
        6: 'June',
        7: 'July',
        8: 'August',
        9: 'September',
        10: 'October',
        11: 'November',
        12: 'December'
    }

    df_global_oil_price_raw = pd.read_csv(file_path)

    date_col = next(
        (
            col for col in df_global_oil_price_raw.columns
            if col.lower() in ['observation_date', 'date']
        ),
        None
    )

    if date_col is None:
        date_col = df_global_oil_price_raw.columns[0]

    price_col = next(
        (
            col for col in df_global_oil_price_raw.columns
            if col.upper() == 'PPOILUSDM'
        ),
        df_global_oil_price_raw.columns[1]
    )

    df_global_oil_price = (
        df_global_oil_price_raw
        .assign(
            Date=lambda x: pd.to_datetime(
                x[date_col], errors='coerce'
            ),
            Global_Oil_Price=lambda x: pd.to_numeric(
                x[price_col], errors='coerce'
            )
        )
        .dropna(subset=['Date', 'Global_Oil_Price'])
        .loc[lambda x: x['Date'].between(
            '2010-01-01', '2026-10-31'
        )]
        .assign(
            Year=lambda x: x['Date'].dt.year,
            Month_number=lambda x: x['Date'].dt.month
        )
        # Agregasi data harian menjadi rata-rata bulanan
        .groupby(
            ['Year', 'Month_number'], as_index=False
        )
        .agg(
            Global_Oil_Price=('Global_Oil_Price', 'mean')
        )
        .assign(
            Month=lambda x: x['Month_number'].map(month_map)
        )
        .sort_values(['Year', 'Month_number'])
        .reset_index(drop=True)
        [['Year', 'Month', 'Month_number', 'Global_Oil_Price']]
    )

    return df_global_oil_price


df_global_oil_price = process_Global_Oil_Price(
    file_path + "DCOILWTICO.csv"
)

print("=== Dataset Information ===")
df_global_oil_price.info()

print("=== Dataset Head ===")
print(df_global_oil_price.head())

print("=== Dataset Tail ===")
print(df_global_oil_price.tail())
```

**Global Price of Rice, Thailand**

*Source* = https://fred.stlouisfed.org/series/pricenpqusdm

```python
def process_Global_Price(file_path, sheet_name = 0):
    month_map = {
        1 : 'January',
        2 : 'February',
        3 : 'March',
        4 : 'April',
        5 : 'May',
        6 : 'June',
        7 : 'July',
        8 : 'August',
        9 : 'September',
        10 : 'October',
        11 : 'November',
        12 : 'December'
    }

    df_global_price_raw = pd.read_excel(file_path, sheet_name=sheet_name)

    date_col = next((col for col in df_global_price_raw.columns if col.lower() in ['observation_date','date']), None)
    if date_col is None:
      date_col = df_global_price_raw.columns[0]

    price_col = next((col for col in df_global_price_raw.columns if col.upper()=='PRICENPQUSDM'), df_global_price_raw.columns[1])

    df_global_price = (
        df_global_price_raw
        .assign(
            Date=lambda x: pd.to_datetime(x[date_col], errors='coerce')
        )
        .dropna(subset=['Date'])
        .assign(
            Year=lambda x: x['Date'].dt.year,
            Month_number=lambda x: x['Date'].dt.month,
            Month=lambda x: x['Date'].dt.month.map(month_map),
            Global_Price=lambda x: pd.to_numeric(x[price_col], errors='coerce')
        )
        .dropna(subset=['Global_Price'])
        .loc[lambda x: x['Date'].between('2010-01-01', '2026-10-31')]
        .sort_values(by=['Year', 'Month_number'])
        .reset_index(drop=True)
        [['Year', 'Month', 'Global_Price']]
    )
    return df_global_price


df_global_price = process_Global_Price(file_path + "Thai Price.xlsx", sheet_name=1)

print("=== Dataset Information ==")
df_global_price.info()

print("=== Dataset Head ==")
df_global_price.head()
```

**Exchange Rate**

*Source*: https://www.bi.go.id/id/statistik/informasi-kurs/transaksi-bi/default.aspx

```python
def process_exchange_rate(file_path):
    month_map = {
        1: 'January',
        2: 'February',
        3: 'March',
        4: 'April',
        5: 'May',
        6: 'June',
        7: 'July',
        8: 'August',
        9: 'September',
        10: 'October',
        11: 'November',
        12: 'December'
    }
    df_exchange_rate_raw = pd.read_excel(file_path + "Kurs Transaksi USD.xlsx", skiprows=4)

    df_exchange_rate = (
        df_exchange_rate_raw
       .assign(
           Date = lambda x:pd.to_datetime(x['Tanggal'], errors = 'coerce'),
           Kurs_Jual=lambda x: pd.to_numeric(x['Kurs Jual'], errors='coerce'),
           Kurs_Beli=lambda x: pd.to_numeric(x['Kurs Beli'], errors='coerce')
       )
        .dropna(subset=['Date', 'Kurs_Jual', 'Kurs_Beli'])
        .assign(
            kurs_rata=lambda x: (x['Kurs_Jual'] + x['Kurs_Beli']) / 2
        )
        .loc[
            lambda x: x['Date'].between('2010-01-01', '2026-10-31')
        ]
        .assign(
            Year=lambda x: x['Date'].dt.year,
            Month_number=lambda x: x['Date'].dt.month,
            Month=lambda x: x['Date'].dt.month.map(month_map)
        )
        .sort_values('Date')
        .reset_index(drop=True)
    )

    df_exchange_rate_monthly = (
        df_exchange_rate
        .groupby(['Year', 'Month','Month_number'],as_index = False, sort = True)['kurs_rata']
        .mean()
        .rename(columns = {'kurs_rata':'Exchange_Rate'})
        .sort_values(by=['Year','Month_number'])
        [['Year','Month','Exchange_Rate']]
    )

    return df_exchange_rate_monthly

df_exchange_rate = process_exchange_rate(file_path)

print("=== Dataset Information ==")
df_exchange_rate.info()

print("=== Dataset Head ==")
df_exchange_rate.head()
```

## Data Processing and Integration

**Data Grouping**

```python
month_map = {
    'January': 1, 'February': 2, 'March': 3, 'April': 4,
    'May': 5, 'June': 6, 'July': 7, 'August': 8,
    'September': 9, 'October': 10, 'November': 11, 'December': 12
}
month_names = {v: k for k, v in month_map.items()}

df_domestic_rice_list = df_domestic_rice_list.rename(columns={'Domestic_Price': 'Domestic_Rice_Price'})
df_geopolitic = df_geopolitic.rename(columns={'GPR': 'GPR_Index'})
df_global_price = df_global_price.rename(columns={'Global_Price': 'Thai_Rice_Price'})
df_oil_price = df_global_oil_price.rename(columns={'Global_Oil_Price': 'Oil_Price'})

datasets = {
    "Domestic Rice": df_domestic_rice_list,
    "GPR": df_geopolitic,
    "Thai Rice": df_global_price,
    "Exchange Rate": df_exchange_rate,
    "Oil Price": df_oil_price
}

def standardisasi_waktu(df, nama):
    df = df.copy()
    df.columns = df.columns.astype(str).str.strip()

    if 'Month_number' not in df.columns:
        if 'Month' not in df.columns:
            raise ValueError(f"{nama}: Kolom 'Month' tidak ditemukan.")
        df['Month_number'] = df['Month'].astype(str).str.strip().map(month_map)

    df['Year'] = pd.to_numeric(df['Year'], errors='coerce')
    df['Month_number'] = pd.to_numeric(df['Month_number'], errors='coerce')

    return (
        df.dropna(subset=['Year', 'Month_number'])
          .astype({'Year': int, 'Month_number': int})
          .query('1 <= Month_number <= 12')
    )

df_full = None

for nama, df in datasets.items():
    df_clean_temp = standardisasi_waktu(df, nama)

    value_cols = [col for col in df_clean_temp.columns if col not in ['Year', 'Month_number', 'Month']]
    col_to_keep = ['Year', 'Month_number'] + value_cols

    df_temp = df_clean_temp[col_to_keep].drop_duplicates(subset=['Year', 'Month_number'])

    if df_full is None:
        df_full = df_temp
    else:
        df_full = pd.merge(df_full, df_temp, on=['Year', 'Month_number'], how='outer', validate='one_to_one')

df_full['Month'] = df_full['Month_number'].map(month_names)

var_cols = [c for c in df_full.columns if c not in ['Year', 'Month', 'Month_number']]
df_full = df_full[['Year', 'Month', 'Month_number'] + var_cols].sort_values(['Year', 'Month_number']).reset_index(drop=True)

df_clean = df_full.dropna(subset=var_cols).reset_index(drop=True)

print("=== FULL DATASET INFO ===")
df_full.info()
print("\n=== 5 DATA TERATAS FULL DATASET ===")
print(df_full.head())

print("\n=== CLEAN DATASET (COMMON PERIOD) INFO ===")
df_clean.info()

print("\n=== RENTANG WAKTU ===")
print(f"Full Dataset   : {df_full['Year'].min()} - {df_full['Year'].max()} (Total: {len(df_full)} baris)")
print(f"Clean Dataset  : {df_clean['Year'].min()} - {df_clean['Year'].max()} (Total: {len(df_clean)} baris)")

print("\n=== JUMLAH MISSING VALUE PADA FULL DATASET ===")
print(df_full[var_cols].isna().sum())

print("\n=== JUMLAH DATA KOSONG BERDASARKAN TAHUN ===")
missing_rows = df_full[df_full[var_cols].isna().any(axis=1)]
missing_by_year = missing_rows.groupby('Year')[var_cols].agg(lambda x: x.isna().sum())
print(missing_by_year)
```

**Columns Addition**

```python
y = df_clean ['GPR_Index'].dropna()

kandidat = np.sort(y.unique())
kandidat_valid = kandidat[int(len(kandidat) * 0.15) : int(len(kandidat) * 0.85)]

min_ssr, threshold_terbaik = float('inf'), None

for gamma in kandidat_valid:
    dummy_rezim = (y > gamma).astype(int)
    model = sm.OLS(y, sm.add_constant(dummy_rezim)).fit()

    if model.ssr < min_ssr:
        min_ssr, threshold_terbaik = model.ssr, gamma

print(f"Threshold SSR untuk GPR_Index: {threshold_terbaik}")
df_clean['dummy_gpr'] = (df_clean['GPR_Index'] > threshold_terbaik).astype(int)

print(df_clean[['GPR_Index', 'dummy_gpr']].head())
df_clean['dummy_gpr'].value_counts()
```

```python
# Verifikasi nilai unik yang ada di dummy_gpr
print("Nilai unik dummy_gpr sebelum dipastikan:")
print(df_clean['dummy_gpr'].unique())
print(df_clean['dummy_gpr'].value_counts())

# Karena dummy_gpr sudah bertipe 0 dan 1 dari langkah threshold,
# kita hanya perlu memastikan tipenya adalah integer secara aman
df_clean['dummy_gpr'] = df_clean['dummy_gpr'].astype(int)

# Verifikasi hasil akhir
print("\nSetelah perbaikan:")
print(df_clean['dummy_gpr'].value_counts())
print("Tipe data:", df_clean['dummy_gpr'].dtype)
```

```python
y = df_clean['ln_Domestic_Rice_Price']
x = df_clean[['GPR_Index','ln_Oil_Price','ln_Thai_Rice_Price','ln_Exchange_Rate']]

# Menggunakan 'dummy_krisis' yang merupakan kolom aktual di df_clean
selected = ardl_select_order(
    endog = y,
    exog = x,
    maxlag = 4,
    maxorder = 4,
    ic='aic',
    trend = 'c',)

print('Lag awal dari ardl_selected_order:')
print('AR lags (p):', selected.ar_lags)
print('Exog lags (q):', selected.dl_lags)

# Memastikan model memiliki variabel eksogen dengan lag >= 1 untuk mencegah error UECM
has_no_exog_or_zero_lag = len(selected.dl_lags) == 0 or any(
    (max(lags) == 0 if isinstance(lags, (list, tuple)) else lags == 0)
    for lags in selected.dl_lags.values()
)

if has_no_exog_or_zero_lag:
    print("\n[INFO] Lag eksogen kosong atau bernilai 0 (UECM membutuhkan lag >= 1 untuk variabel eksogen).")
    print("[INFO] Beralih ke Grid Search manual dengan rentang lag (1 s.d. 4)...\n")

    candidates = []
    for p_lag in range(1, 5):
        for q_lag in range(1, 5):
            try:
                model = ARDL(
                    endog=y,
                    lags=p_lag,
                    exog=x,
                    order=q_lag,
                    trend='c',
                    missing='drop',
                ).fit()
                candidates.append({
                    'AIC': model.aic,
                    'p': p_lag,
                    'q': q_lag,
                    'model': model
                })
            except Exception:
                continue

    candidates.sort(key=lambda z: z['AIC'])
    best_spec = candidates[0]

    final_fit = best_spec['model']
    final_ardl_model = final_fit.model
    print(f"Optimal ARDL Order terpilih (p, q): ({best_spec['p']}, {best_spec['q']})")
    print(f"AIC Terendah: {final_fit.aic:.4f}")

else:
    print("\n[INFO] Semua lag eksogen >= 1. Menggunakan hasil ardl_select_order.")
    final_fit = selected.model.fit()
    final_ardl_model = selected.model

print("\n=== ARDL Estimation Summary ===")
print(final_fit.summary())

uecm_model = UECM.from_ardl(final_ardl_model)
uecm_res = uecm_model.fit()

print("\n=== UECM Estimation Summary ===")
print(uecm_res.summary())

bounds_res = uecm_res.bounds_test(case=3)

print("\n=== ARDL Bounds Test Results ===")
print(bounds_res)

# ==================================================
# 4. KOEFISIEN JANGKA PANJANG & SPEED OF ADJUSTMENT
# ==================================================
params = uecm_res.params
pvalues = uecm_res.pvalues

y_name = y.name if hasattr(y, 'name') else 'ln_Domestic_Rice_Price'
speed_of_adj = params[f'{y_name}.L1']
speed_pval = pvalues[f'{y_name}.L1']

print("\n=== SPEED OF ADJUSTMENT (EC_t-1) ===")
print(f"Koefisien EC_t-1 : {speed_of_adj:.4f}")
print(f"p-value          : {speed_pval:.4f}")

long_run_data = []
for col in x.columns:
    pi_x = params[f'{col}.L1']
    lr_coef = - pi_x / speed_of_adj  # Normalisasi: -\pi_x / \pi_y
    long_run_data.append({
        'Variable': col,
        'Long-Run Coef': round(lr_coef, 4),
        'Unrestricted L1 Coef': round(pi_x, 4),
        'p-value (L1)': round(pvalues[f'{col}.L1'], 4)
    })

print("\n=== NORMALIZED LONG-RUN COEFFICIENTS ===")
print(pd.DataFrame(long_run_data).to_string(index=False))

# ==================================================
# 5. DINAMIKA JANGKA PENDEK
# ==================================================
short_run_data = []
for var_name, coef in params.items():
    if var_name.startswith('D.') or 'crisis_dummy' in var_name or 'dummy_krisis' in var_name:
        short_run_data.append({
            'Variable (Short-Run)': var_name,
            'Coefficient': round(coef, 4),
            'Std Error': round(uecm_res.bse[var_name], 4),
            't-statistic': round(uecm_res.tvalues[var_name], 4),
            'p-value': round(pvalues[var_name], 4)
        })

print("\n=== KOEFISIEN DINAMIKA JANGKA PENDEK ===")
print(pd.DataFrame(short_run_data).to_string(index=False))
```

**Structural Break Test**
1. Quandt-Andrews (Sup-F)

```python
import sys
import subprocess
import numpy as np
import pandas as pd
import statsmodels.api as sm

# Auto-install library ruptures jika belum terinstall
try:
    import ruptures as rpt
except ModuleNotFoundError:
    print("[INFO] Installing missing dependency 'ruptures'...")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "ruptures"])
    import ruptures as rpt

# ==================================================
# 1. PERSIAPAN MATRIKS REGRESI (ARDL Level Specification)
# ==================================================
reg = pd.DataFrame(index=df_clean.index)

reg["const"] = 1.0
reg["y_l1"] = y.shift(1)
reg["y_l2"] = y.shift(2)

for col in x.columns:
    reg[col + "_l0"] = x[col]
    reg[col + "_l1"] = x[col].shift(1)

reg["y"] = y
reg = reg.dropna()

base_X = reg.drop(columns="y")
base = sm.OLS(reg["y"], base_X).fit()

# ==================================================
# 2. QUANDT-ANDREWS / SUP-F SCAN (SINGLE BREAK)
# ==================================================
scan_results = []

# Trimming 15% awal dan 85% akhir
start = int(len(reg) * 0.15)
end = int(len(reg) * 0.85)

for idx in range(start, end):
    break_date = reg.index[idx]
    D = (reg.index >= break_date).astype(int)

    Z = base_X.copy()
    Z["break_dummy"] = D

    interaction_terms = []
    for col in base_X.columns:
        if col != "const":
            name = col + "_post"
            Z[name] = base_X[col] * D  # Perkalian regressors dengan dummy D
            interaction_terms.append(name)

    terms = ["break_dummy"] + interaction_terms
    R = np.zeros((len(terms), Z.shape[1]))

    for i, term in enumerate(terms):
        R[i, Z.columns.get_loc(term)] = 1

    fit = sm.OLS(reg["y"], Z).fit()
    test = fit.f_test(R)

    scan_results.append({
        "break_date": break_date,
        "F_stat": float(test.fvalue),
        "p_value_uncorrected": float(test.pvalue),
        "R2": fit.rsquared,
        "model": fit
    })

break_table = pd.DataFrame([
    {k: v for k, v in row.items() if k != "model"}
    for row in scan_results
])

best_break = max(scan_results, key=lambda z: z["F_stat"])

print("=== QUANDT-ANDREWS / SUP-F SCAN RESULTS ===")
print("\nTOP 10 BREAK CANDIDATES:")
print(
    break_table.sort_values("F_stat", ascending=False)
    .head(10).to_string(index=False)
)

print(f"\nSelected break (Sup-F Max) : {best_break['break_date']}")
print(f"Sup-F Statistic            : {best_break['F_stat']:.4f}")
print(f"Uncorrected p-value        : {best_break['p_value_uncorrected']:.6e}")

# ==================================================
# 3. BAI-PERRON MULTIPLE BREAKS (RUPTURES)
# ==================================================
data_mat = reg.values
min_size = int(len(reg) * 0.15)  # Trimming 15%

print("\n=== BAI-PERRON MULTIPLE BREAK RESULTS (ruptures) ===")

# --- METODE A: PELT (Otomatis menentukan jumlah break optimal via Penalti) ---
algo_pelt = rpt.Pelt(model="linear", min_size=min_size).fit(data_mat)
pelt_indices = algo_pelt.predict(pen=5)  # Parameter penalti (umumnya 5 - 15)
pelt_dates = [reg.index[idx - 1] for idx in pelt_indices[:-1]]

print("\n1. Otomatis (PELT Method - Penalty = 5):")
print(f"   Jumlah Break Optimal Terdeteksi: {len(pelt_dates)}")
for i, date in enumerate(pelt_dates, 1):
    print(f"   - Break {i}: {date.strftime('%Y-%m-%d')}")

# --- METODE B: Dynamic Programming (Menentukan jumlah break spesifik, misal max 5) ---
max_breaks = 5
algo_dynp = rpt.Dynp(model="linear", min_size=min_size).fit(data_mat)
dynp_indices = algo_dynp.predict(n_bkps=max_breaks)
dynp_dates = [reg.index[idx - 1] for idx in dynp_indices[:-1]]

print(f"\n2. Jumlah Break Ditentukan (Dynamic Programming - Max {max_breaks} Breaks):")
print(f"   Jumlah Break: {len(dynp_dates)}")
for i, date in enumerate(dynp_dates, 1):
    print(f"   - Break {i}: {date.strftime('%Y-%m-%d')}")
```

```python
import pandas as pd
import numpy as np
from statsmodels.tsa.ardl import ARDL, ardl_select_order, UECM

# ==================================================
# 1. SIAPKAN DUMMY KRISIS (PULSE DUMMY)
# ==================================================
df_clean['crisis_dummy'] = 0

crisis_dates = [    '2020-04-01', '2020-05-01', '2020-06-01', '2020-07-01',
    '2020-08-01', '2020-09-01', '2020-10-01', '2020-11-01',
    '2020-12-01', '2021-01-01',
]

df_clean.loc[df_clean.index.strftime('%Y-%m-%d').isin(crisis_dates), 'crisis_dummy'] = 1

y = df_clean['ln_Domestic_Rice_Price']
x = df_clean[['GPR_Index', 'ln_Exchange_Rate', 'ln_Thai_Rice_Price','ln_Oil_Price']]
fixed_dummy = df_clean[['crisis_dummy']]

# ==================================================
# 2. SELEKSI LAG OPTIMAL & FIT MODEL ARDL
# ==================================================
selected = ardl_select_order(
    endog=y,
    exog=x,
    maxlag=4,
    maxorder=4,
    ic='aic',
    trend='c'
)

# Cek apakah ada lag 0 pada eksogen (mencegah error UECM)
has_zero_lag = any(
    (max(lags) == 0 if isinstance(lags, (list, tuple)) else lags == 0)
    for lags in selected.dl_lags.values()
)

if has_zero_lag:
    print("[INFO] Ditemukan lag 0 pada eksogen. Menjalankan Grid Search (q >= 1)...")
    candidates = []
    for p_lag in range(1, 5):
        for q_lag in range(1, 5):
            try:
                model = ARDL(
                    endog=y,
                    lags=p_lag,
                    exog=x,
                    order=q_lag,
                    fixed=fixed_dummy,
                    trend='c',
                    missing='drop'
                ).fit()
                candidates.append({'AIC': model.aic, 'model': model})
            except Exception:
                continue
    candidates.sort(key=lambda z: z['AIC'])
    final_fit = candidates[0]['model']
else:
    final_ardl = ARDL(
        endog=y,
        lags=selected.ar_lags,
        exog=x,
        order=selected.dl_lags,
        fixed=fixed_dummy,
        trend='c'
    )
    final_fit = final_ardl.fit()

print("=== SUMMARY ARDL MODEL ===")
print(final_fit.summary())

# ==================================================
# 3. ESTIMASI UECM & UJI KOINTEGRASI (BOUNDS TEST)
# ==================================================
uecm_model = UECM.from_ardl(final_fit.model)
uecm_res = uecm_model.fit()

print("\n=== SUMMARY UECM MODEL ===")
print(uecm_res.summary())

bounds_res = uecm_res.bounds_test(case=3)
print("\n=== ARDL BOUNDS TEST (UJI KOINTEGRASI) ===")
print(bounds_res)

# ==================================================
# 4. KOEFISIEN JANGKA PANJANG & SPEED OF ADJUSTMENT
# ==================================================
params = uecm_res.params
pvalues = uecm_res.pvalues

y_name = y.name if hasattr(y, 'name') else 'ln_Domestic_Rice_Price'
speed_of_adj = params[f'{y_name}.L1']
speed_pval = pvalues[f'{y_name}.L1']

print("\n=== SPEED OF ADJUSTMENT (EC_t-1) ===")
print(f"Koefisien EC_t-1 : {speed_of_adj:.4f}")
print(f"p-value          : {speed_pval:.4f}")

long_run_data = []
for col in x.columns:
    pi_x = params[f'{col}.L1']
    lr_coef = - pi_x / speed_of_adj  # Normalisasi: -\pi_x / \pi_y
    long_run_data.append({
        'Variable': col,
        'Long-Run Coef': round(lr_coef, 4),
        'Unrestricted L1 Coef': round(pi_x, 4),
        'p-value (L1)': round(pvalues[f'{col}.L1'], 4)
    })

print("\n=== NORMALIZED LONG-RUN COEFFICIENTS ===")
print(pd.DataFrame(long_run_data).to_string(index=False))

# ==================================================
# 5. DINAMIKA JANGKA PENDEK
# ==================================================
short_run_data = []
for var_name, coef in params.items():
    if var_name.startswith('D.') or 'crisis_dummy' in var_name:
        short_run_data.append({
            'Variable (Short-Run)': var_name,
            'Coefficient': round(coef, 4),
            'Std Error': round(uecm_res.bse[var_name], 4),
            't-statistic': round(uecm_res.tvalues[var_name], 4),
            'p-value': round(pvalues[var_name], 4)
        })

print("\n=== KOEFISIEN DINAMIKA JANGKA PENDEK ===")
print(pd.DataFrame(short_run_data).to_string(index=False))
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from statsmodels.tsa.ardl import ARDL, ardl_select_order, UECM

# ==================================================
# 1. FUNGSI DEKOMPOSISI NARDL (PARTIAL SUMS)
# ==================================================
def decompose_nardl(series):
    """
    Memecah variabel menjadi akumulasi perubahan positif (POS)
    dan negatif (NEG) sesuai metode Shin et al. (2014).
    """
    delta = series.diff()

    # Akumulasi shock positif (kenaikan)
    pos = delta.clip(lower=0).cumsum().fillna(0)
    # Akumulasi shock negatif (penurunan)
    neg = delta.clip(upper=0).cumsum().fillna(0)

    # Menambahkan nilai observasi awal sebagai base
    base = series.iloc[0] if pd.notna(series.iloc[0]) else 0
    return pos + base, neg + base

# Terapkan dekomposisi pada variabel utama
df_nardl = df_clean.copy()

# thai_pos = Kenaikan Harga Thailand | thai_neg = Penurunan Harga Thailand
df_nardl['thai_pos'], df_nardl['thai_neg'] = decompose_nardl(df_nardl['ln_Thai_Rice_Price'])

# er_pos = Depresiasi Rupiah (Kurs Naik) | er_neg = Apresiasi Rupiah (Kurs Turun)
df_nardl['er_pos'], df_nardl['er_neg'] = decompose_nardl(df_nardl['ln_Exchange_Rate'])

df_nardl['op_pos'], df_nardl['op_neg'] = decompose_nardl(df_nardl['ln_Oil_Price'])

# Hapus missing values akibat proses differencing di awal
df_nardl = df_nardl.dropna()

# ==================================================
# 2. SELEKSI LAG OPTIMAL & FIT MODEL NARDL
# ==================================================
y_nardl = df_nardl['ln_Domestic_Rice_Price']

# Memasukkan komponen POS dan NEG, serta kontrol simetris (GPR_Index & ln_Oil_Price)
x_nardl = df_nardl[['thai_pos', 'thai_neg', 'er_pos', 'er_neg', 'dummy_gpr', 'op_pos','op_neg' ]]

# Seleksi lag optimal (TANPA fixed regressor / dummy)
selected_nardl = ardl_select_order(
    endog=y_nardl,
    exog=x_nardl,
    maxlag=4,
    maxorder=4,
    ic='aic',
    trend='c',
    fixed =df_clean['dummy_gpr']
)

# Pengecekan lag 0 (Mencegah error UECM)
has_zero_lag = any(
    (max(lags) == 0 if isinstance(lags, (list, tuple)) else lags == 0)
    for lags in selected_nardl.dl_lags.values()
)

if has_zero_lag:
    print("[INFO] Ditemukan lag 0 pada eksogen NARDL. Menjalankan Grid Search (q >= 1)...")
    candidates = []
    for p_lag in range(1, 5):
        for q_lag in range(1, 5):
            try:
                model = ARDL(
                    endog=y_nardl,
                    lags=p_lag,
                    exog=x_nardl,
                    order=q_lag,
                    trend='c',
                    missing='drop'
                ).fit()
                candidates.append({'AIC': model.aic, 'model': model})
            except Exception:
                continue
    candidates.sort(key=lambda z: z['AIC'])
    final_nardl_fit = candidates[0]['model']
else:
    final_nardl_model = ARDL(
        endog=y_nardl,
        lags=selected_nardl.ar_lags,
        exog=x_nardl,
        order=selected_nardl.dl_lags,
        trend='c'
    )
    final_nardl_fit = final_nardl_model.fit()

print("=== SUMMARY NARDL MODEL ===")
print(final_nardl_fit.summary())

# ==================================================
# 3. ESTIMASI UECM (ERROR CORRECTION MODEL)
# ==================================================
uecm_nardl = UECM.from_ardl(final_nardl_fit.model)
uecm_nardl_res = uecm_nardl.fit()

print("\n=== SUMMARY UECM (NARDL) MODEL ===")
print(uecm_nardl_res.summary())

# Bounds Test untuk Kointegrasi Model Asimetris
bounds_res_nardl = uecm_nardl_res.bounds_test(case=3)
print("\n=== NARDL BOUNDS TEST (UJI KOINTEGRASI) ===")
print(bounds_res_nardl)

# ==================================================
# 4. LONG-RUN COEFFICIENTS & WALD TEST (UJI ASIMETRI)
# ==================================================
params = uecm_nardl_res.params
pvalues = uecm_nardl_res.pvalues

y_name = y_nardl.name if hasattr(y_nardl, 'name') else 'ln_Domestic_Rice_Price'
speed_of_adj = params[f'{y_name}.L1']

print("\n=== SPEED OF ADJUSTMENT (EC_t-1) ===")
print(f"Koefisien EC_t-1 : {speed_of_adj:.4f} (p-value: {pvalues[f'{y_name}.L1']:.4f})")

long_run_nardl = []
for col in x_nardl.columns:
    key_name = f'{col}.L1'
    if key_name in params:
        lr_coef = - params[key_name] / speed_of_adj
        long_run_nardl.append({
            'Variabel': col,
            'Long-Run Coef': round(lr_coef, 4),
            'p-value (Unrestricted L1)': round(pvalues[key_name], 4)
        })

print("\n=== NORMALIZED LONG-RUN COEFFICIENTS (NARDL) ===")
print(pd.DataFrame(long_run_nardl).to_string(index=False))

# --- WALD TEST UNTUK ASIMETRI JANGKA PANJANG ---
# Hipotesis Nol (H0): Koefisien POS = Koefisien NEG (Simetris)
# Jika p-value < 0.05, Tolak H0 -> Terdapat Asimetri Jangka Panjang

print("\n=== UJI WALD: ASIMETRI JANGKA PANJANG ===")
try:
    # Uji Asimetri Harga Beras Thailand
    wald_thai = uecm_nardl_res.wald_test("thai_pos.L1 = thai_neg.L1")
    print(f"Harga Thai (POS vs NEG) -> F-stat: {wald_thai.statistic:.4f}, p-value: {wald_thai.pvalue:.4f}")
except Exception as e:
    print("Wald Test Harga Thai gagal dihitung:", e)

try:
    # Uji Asimetri Nilai Tukar Rupiah
    wald_er = uecm_nardl_res.wald_test("er_pos.L1 = er_neg.L1")
    print(f"Nilai Tukar (POS vs NEG) -> F-stat: {wald_er.statistic:.4f}, p-value: {wald_er.pvalue:.4f}")
except Exception as e:
    print("Wald Test Nilai Tukar gagal dihitung:", e)
```

**Objective 3 - Early Warning Indicator Exploration**

Mengeksplorasi apakah kombinasi indikator harga internasional, nilai tukar, dan risiko geopolitik dapat memberikan informasi tambahan untuk memantau potensi tekanan harga beras domestik pada periode mendatang.

Analisis diarahkan pada identifikasi indikator yang relevan, pengujian kemampuan prediksi, dan penyusunan rancangan awal kerangka pemantauan risiko.

Expected output: Dashboard pemantauan indikator eksternal, matriks kondisi risiko berdasarkan kriteria yang terukur, serta evaluasi awal kemampuan indikator dalam memprediksi perubahan harga domestik.

Matriks tersebut merupakan prototipe analitis, bukan sistem peringatan dini operasional yang telah tervalidasi.

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# ==================================================
# 1. DEFINISI TARGET & FITUR (PRICE SPIKE)
# ==================================================
df_ewi = df_clean.copy()

# Menghitung volatilitas/pertumbuhan bulanan (first difference)
df_ewi['d_DomPrice'] = df_ewi['ln_Domestic_Rice_Price'].diff()
df_ewi['d_ThaiPrice'] = df_ewi['ln_Thai_Rice_Price'].diff()
df_ewi['d_ER'] = df_ewi['ln_Exchange_Rate'].diff()
df_ewi['d_GPR'] = df_ewi['GPR_Index'].diff()
df_ewi['d_Oil'] = df_ewi['ln_Oil_Price'].diff()

df_ewi = df_ewi.dropna()

# Definisi Spike: Lonjakan harga domestik melebihi rata-rata + 0.5 Standar Deviasi
threshold = df_ewi['d_DomPrice'].mean() + (0.5 * df_ewi['d_DomPrice'].std())
df_ewi['Spike_Signal'] = (df_ewi['d_DomPrice'] >= threshold).astype(int)

# ==================================================
# 2. EVALUASI MODEL PROBIT (DAN LOGIT/OLS)
# ==================================================
# Menggunakan variabel dummy_gpr bersama indikator eksternal lainnya (lag-1)
X_cols = ['d_ThaiPrice', 'd_ER', 'dummy_gpr','d_Oil']
X_lag = df_ewi[X_cols].shift(1).dropna()
y_target = df_ewi['Spike_Signal'].loc[X_lag.index]

X_model = sm.add_constant(X_lag)

try:
    # Menggunakan model Probit
    ews_model = sm.Probit(y_target, X_model).fit(disp=0)
    model_type = "PROBIT"
except Exception as e:
    try:
        # Fallback ke Logit jika Probit gagal konvergen
        ews_model = sm.Logit(y_target, X_model).fit(disp=0)
        model_type = "LOGIT"
    except Exception as e2:
        # Fallback terakhir ke OLS (Linear Probability Model)
        ews_model = sm.OLS(y_target, X_model).fit()
        model_type = "OLS (Linear Probability)"

print(f"=== SUMMARY EWS MODEL ({model_type}) ===")
print(ews_model.summary())

# Menyimpan hasil probabilitas prediksi ke dataframe
df_ewi.loc[X_lag.index, 'Probabilitas_Spike'] = ews_model.predict(X_model)

# ==================================================
# 3. PENDEKATAN SIGNAL (KAMINSKY-LIZONDO APPROACH)
# ==================================================
# Menghitung persentase sinyal benar vs false alarm dari dummy_gpr
df_ewi['Signal_GPR'] = df_ewi['dummy_gpr'].shift(1) # Menggunakan lag-1

# Matriks sederhana untuk melihat efektivitas Signal_GPR terhadap Spike_Signal
crosstab = pd.crosstab(df_ewi['Signal_GPR'], df_ewi['Spike_Signal'], normalize='index')
print("\n=== EFEKTIVITAS SIGNAL GPR (KAMINSKY-LIZONDO RATIO) ===")
print(crosstab)

# Skor Risiko Dini gabungan (0 = Aman, 3 = Sangat Berisiko)
# Mengubah variabel d_ThaiPrice dan d_ER menjadi flag (> mean + std)
for col in ['d_ThaiPrice', 'd_ER']:
    std_val = df_ewi[col].std()
    mean_val = df_ewi[col].mean()
    df_ewi[f'Flag_{col}'] = (df_ewi[col] > (mean_val + std_val)).astype(int)

df_ewi['Risk_Score'] = (df_ewi['Flag_d_ThaiPrice'] + df_ewi['Flag_d_ER'] + df_ewi['dummy_gpr']).shift(1)

def assign_risk_status(score):
    if score >= 2: return "Tinggi (Merah)"
    elif score == 1: return "Waspada (Kuning)"
    elif score == 0: return "Normal (Hijau)"
    else: return "N/A"

df_ewi['Status_Risiko'] = df_ewi['Risk_Score'].apply(assign_risk_status)

print("\n=== MATRIKS KONDISI RISIKO (10 PERIODE TERAKHIR) ===")
display_cols = ['d_DomPrice', 'dummy_gpr', 'Probabilitas_Spike', 'Risk_Score', 'Status_Risiko']
print(df_ewi[display_cols].tail(10))

# ==================================================
# 4. VISUALISASI DASHBOARD EWS
# ==================================================
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 10), dpi=120, sharex=True)

# --- PLOT 1: Harga Domestik & Area Risiko ---
ax1.plot(df_ewi.index, df_ewi['ln_Domestic_Rice_Price'], color='#2b2b2b', linewidth=2, label='ln(Harga Beras Domestik)')

for i in range(1, len(df_ewi)):
    start = df_ewi.index[i-1]
    end = df_ewi.index[i]
    score = df_ewi['Risk_Score'].iloc[i]

    if score >= 2:
        ax1.axvspan(start, end, color='#ff4d4d', alpha=0.3, lw=0)  # Merah
    elif score == 1:
        ax1.axvspan(start, end, color='#ffcc00', alpha=0.3, lw=0)  # Kuning

spike_dates = df_ewi[df_ewi['Spike_Signal'] == 1].index
ax1.scatter(spike_dates, df_ewi.loc[spike_dates, 'ln_Domestic_Rice_Price'], color='red', zorder=5, label='Actual Price Spike')

ax1.set_title('Dashboard EWS: Harga Beras Domestik vs Status Risiko (Berbasis Dummy GPR & Indikator)', fontsize=13, fontweight='bold')
ax1.set_ylabel('ln(Harga Domestik)')
ax1.grid(True, linestyle='--', alpha=0.5)
ax1.legend(loc='upper left')

# --- PLOT 2: Probabilitas Prediksi Model ---
ax2.plot(df_ewi.index, df_ewi['Probabilitas_Scored'] if 'Probabilitas_Scored' in df_ewi else df_ewi['Probabilitas_Spike'], color='#0275d8', linewidth=2, label='Probabilitas Tekanan (Model Probit)')
ax2.axhline(y=0.5, color='red', linestyle='--', label='Batas Kritis (50%)')
ax2.fill_between(df_ewi.index, df_ewi['Probabilitas_Spike'], 0, color='#0275d8', alpha=0.1)

ax2.set_title('Probabilitas Terjadinya Lonjakan Harga (Model Probit)', fontsize=13, fontweight='bold')
ax2.set_ylabel('Probabilitas')
ax2.set_xlabel('Tahun')
ax2.grid(True, linestyle='--', alpha=0.5)
ax2.legend(loc='upper left')

plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m'))
plt.gca().xaxis.set_major_locator(mdates.YearLocator(2))
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
 import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# ==================================================
# 1. PERSIAPAN DATA & PERHITUNGAN VOLATILITAS
# ==================================================
df_ews = df_clean.copy()

# Menghitung persentase perubahan bulanan (first difference dari log)
df_ews['d_DomPrice'] = df_ews['ln_Domestic_Rice_Price'].diff()
df_ews['d_ThaiPrice'] = df_ews['ln_Thai_Rice_Price'].diff()
df_ews['d_ER'] = df_ews['ln_Exchange_Rate'].diff()
df_ews['d_GPR'] = df_ews['GPR_Index'].diff()
df_ews['d_Oil'] = df_ews['ln_Oil_Price'].diff()

df_ews = df_ews.dropna()

# ==================================================
# 2. SISTEM MATRIKS HEURISTIK (SCORING)
# ==================================================
# Ambang batas (Threshold) ditetapkan sebesar: Rata-rata + 1 Standar Deviasi
indicators = ['d_ThaiPrice', 'd_ER', 'dummy_gpr', 'd_Oil']
flag_cols = []

for col in indicators:
    mean_val = df_ews[col].mean()
    std_val = df_ews[col].std()
    threshold = mean_val + std_val

    flag_name = f'Flag_{col}'
    flag_cols.append(flag_name)
    # Beri nilai 1 jika kenaikan melebihi batas toleransi historis
    df_ews[flag_name] = (df_ews[col] > threshold).astype(int)

# Menghitung Total Skor Risiko (Akumulasi Flag pada periode berjalan)
df_ews['Total_Pressure'] = df_ews[flag_cols].sum(axis=1)

# EWS bersifat leading indicator: Tekanan bulan lalu menjadi Peringatan bulan sini
df_ews['Risk_Score'] = df_ews['Total_Pressure'].shift(1).fillna(0)

# Kategori Status Risiko
def assign_risk_status(score):
    if score >= 2: return "Tinggi (Merah)"
    elif score == 1: return "Waspada (Kuning)"
    else: return "Normal (Hijau)"

df_ews['Status_Risiko'] = df_ews['Risk_Score'].apply(assign_risk_status)

# ==================================================
# 3. CETAK MATRIKS KONDISI RISIKO
# ==================================================
print("\n=== MATRIKS KONDISI RISIKO (12 PERIODE TERAKHIR) ===")
display_df = df_ews[['ln_Domestic_Rice_Price', 'Risk_Score', 'Status_Risiko']].copy()
display_df['Triggers_From_Last_Month'] = df_ews.apply(
    lambda row: ', '.join([
        col.replace('Flag_d_', '').replace('Flag_', '')
        for col in flag_cols if df_ews[col].shift(1).loc[row.name] == 1
    ]) or "None", axis=1
)

print(display_df.tail(12).to_string())

# ==================================================
# 4. VISUALISASI DASHBOARD EARLY WARNING SYSTEM
# ==================================================
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 11), dpi=120, sharex=True, gridspec_kw={'height_ratios': [2, 1]})

# --- PANEL 1: Pergerakan Harga Domestik & Shading Status Risiko ---
ax1.plot(df_ews.index, df_ews['ln_Domestic_Rice_Price'], color='#2b2b2b', linewidth=2.5, label='ln(Harga Beras Domestik)')

for i in range(1, len(df_ews)):
    start = df_ews.index[i-1]
    end = df_ews.index[i]
    score = df_ews['Risk_Score'].iloc[i]

    if score >= 2:
        ax1.axvspan(start, end, color='#d9534f', alpha=0.3, lw=0)  # Merah
    elif score == 1:
        ax1.axvspan(start, end, color='#f0ad4e', alpha=0.3, lw=0)  # Kuning

ax1.axvspan(df_ews.index[0], df_ews.index[0], color='#d9534f', alpha=0.3, label='Risiko Tinggi (>=2 Pemicu)')
ax1.axvspan(df_ews.index[0], df_ews.index[0], color='#f0ad4e', alpha=0.3, label='Waspada (1 Pemicu)')

ax1.set_title('Dashboard EWS: Harga Beras Domestik & Status Tekanan Eksternal', fontsize=14, fontweight='bold', pad=10)
ax1.set_ylabel('ln(Harga Beras Domestik)')
ax1.grid(True, linestyle='--', alpha=0.4)
ax1.legend(loc='upper left', frameon=True, facecolor='white', framealpha=0.9)

# --- PANEL 2: Stacked Bar Indikator Pemicu (Triggers) TERMASUK OIL ---
bar_data = df_ews[flag_cols].shift(1).fillna(0)
width = 20  # lebar bar dalam hitungan hari

# Menumpuk 4 variabel secara berurutan
p1 = ax2.bar(df_ews.index, bar_data['Flag_d_ThaiPrice'], width, color='#0275d8', label='Harga Thai')
p2 = ax2.bar(df_ews.index, bar_data['Flag_d_ER'], width, bottom=bar_data['Flag_d_ThaiPrice'], color='#5cb85c', label='Depresiasi Rupiah')
p3 = ax2.bar(df_ews.index, bar_data['Flag_dummy_gpr'], width, bottom=bar_data['Flag_d_ThaiPrice'] + bar_data['Flag_d_ER'], color='#5bc0de', label='Risiko Geopolitik (GPR)')
p4 = ax2.bar(df_ews.index, bar_data['Flag_d_Oil'], width, bottom=bar_data['Flag_d_ThaiPrice'] + bar_data['Flag_d_ER'] + bar_data['Flag_dummy_gpr'], color='#f0ad4e', label='Harga Minyak (Oil)')

ax2.set_title('Komposisi Pemicu Tekanan (Bulan Sebelumnya)', fontsize=12, fontweight='bold', pad=10)
ax2.set_ylabel('Jumlah Flag')
ax2.set_yticks([0, 1, 2, 3, 4])
ax2.grid(axis='y', linestyle='--', alpha=0.4)
ax2.legend(loc='upper left', ncol=4, frameon=True, facecolor='white', framealpha=0.9)

# Formatting Axis Waktu
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m'))
plt.gca().xaxis.set_major_locator(mdates.YearLocator(1))
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```
